```python
import os
import getpass
from google.colab import files

from pypdf import PdfReader

from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate

print("Libraries loaded successfully!")
```

```python
os.environ["GROQ_API_KEY"] = getpass.getpass(
    "Enter your Groq API key: "
)

print("Groq API key added successfully!")
```

```python
uploaded = files.upload()
```

```python
resume_filename = list(uploaded.keys())[0]

reader = PdfReader(resume_filename)

resume_text = ""

for page in reader.pages:
    text = page.extract_text()

    if text:
        resume_text += text + "\n"

print("Resume loaded successfully!")
print("Characters extracted:", len(resume_text))
```

```python
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0.2
)

print("Groq model ready!")
```

```python
job_description = """
We are looking for an AI/ML Engineer.

Required skills:
Python
Machine Learning
Deep Learning
SQL
TensorFlow
Data Analysis
Generative AI
"""

print(job_description)
```

```python
prompt = ChatPromptTemplate.from_template("""
You are CareerMate AI, an intelligent career assistant.

Your task is to analyze the student's resume against the provided job description.

RESUME:
{resume}

JOB DESCRIPTION:
{job_description}

IMPORTANT:
You MUST provide ALL 7 sections below.
Do not skip any section.
Use the exact headings given below.
Section 7 is mandatory even if you cannot access live job listings.

============================

1. RESUME SUMMARY

Give a short and simple summary of the candidate's education,
experience, projects, and overall technical profile.

============================

2. SKILLS FOUND

List the technical skills and soft skills found in the resume.

============================

3. MATCHING SKILLS

List the skills that are present in BOTH:
- the student's resume
- the job description

============================

4. MISSING SKILLS

List the important skills required by the job description
that are NOT clearly present in the student's resume.

============================

5. INTERVIEW QUESTIONS

Generate exactly 5 technical interview questions based on:
- the student's resume
- the job description

============================

6. LEARNING ROADMAP

Create a simple 30-day learning roadmap for the missing skills.

Divide it into:
- Week 1
- Week 2
- Week 3
- Week 4

============================

7. JOB SEARCH & APPLICATION GUIDANCE

This section is MANDATORY.

Based on the job description and the candidate's skills:

A. TARGET JOB ROLES
Suggest 5 job titles that match the job description.

B. WHERE TO FIND JOBS
Suggest suitable platforms where the candidate can search
for these jobs, such as:
- LinkedIn
- Naukri
- Indeed
- Wellfound
- company career pages

C. SEARCH KEYWORDS
Give specific keywords the candidate can type into job portals.

D. APPLICATION TIPS
Give 5 simple and practical tips for applying to these jobs.

E. JOB SEARCH STRATEGY
Explain step-by-step how the candidate can search for,
shortlist, and apply to suitable jobs.

Do NOT invent specific currently available job openings.
This section should provide job-search guidance rather than
claiming that a specific job is currently available.

============================

FINAL INSTRUCTION:

Always display all 7 sections.

Keep the explanation simple, clear, student-friendly,
and useful for a college student preparing for a career.
""")
```

```python
chain = prompt | llm

print("CareerMate AI chain created successfully!")
```

```python
response = chain.invoke({
    "resume": resume_text,
    "job_description": job_description
})

print("\n========== CAREERMATE AI ==========\n")
print(response.content)
```
