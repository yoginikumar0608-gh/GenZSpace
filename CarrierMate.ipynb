```python
import os
import getpass
from google.colab import files

from pypdf import PdfReader

from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate

print("Libraries loaded successfully!")
```

```python
os.environ["GROQ_API_KEY"] = getpass.getpass(
    "Enter your Groq API key: "
)

print("Groq API key added successfully!")
```

```python
uploaded = files.upload()
```

```python
resume_filename = list(uploaded.keys())[0]

reader = PdfReader(resume_filename)

resume_text = ""

for page in reader.pages:
    text = page.extract_text()

    if text:
        resume_text += text + "\n"

print("Resume loaded successfully!")
print("Characters extracted:", len(resume_text))
```

```python
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0.2
)

print("Groq model ready!")
```

```python
job_description = """
We are looking for an AI/ML Engineer.

Required skills:
Python
Machine Learning
Deep Learning
SQL
TensorFlow
Data Analysis
Generative AI
"""

print(job_description)
```

```python
prompt = ChatPromptTemplate.from_template("""
You are CareerMate AI, an intelligent career assistant.

Analyze the student's resume against the job description.

RESUME:
{resume}

JOB DESCRIPTION:
{job_description}

IMPORTANT RULES:
- You MUST provide ALL 7 sections.
- NEVER skip Section 7.
- Use the exact headings shown below.
- Keep the answer simple and student-friendly.
- Do not invent information that is not present in the resume.
- Do not claim that a specific job opening currently exists.
- Section 7 should provide job-search guidance, not fake job listings.

==================================================

1. RESUME SUMMARY

Give a short summary of the candidate including:
- Education
- Experience
- Projects
- Technical profile

==================================================

2. SKILLS FOUND

List the skills found in the resume.

Separate them into:

Technical Skills:
- ...

Soft Skills:
- ...

==================================================

3. MATCHING SKILLS

List the skills that appear in BOTH the resume
and the job description.

Also provide:

MATCHING SKILLS COUNT: [number]

==================================================

4. MISSING SKILLS

List important skills required by the job description
that are not clearly present in the resume.

Also provide:

MISSING SKILLS COUNT: [number]

==================================================

5. INTERVIEW QUESTIONS

Generate exactly 5 technical interview questions
based on the student's resume and the job description.

==================================================

6. 30-DAY LEARNING ROADMAP

Create a simple 30-day roadmap for the missing skills.

WEEK 1:
...

WEEK 2:
...

WEEK 3:
...

WEEK 4:
...

==================================================

7. JOB SEARCH & APPLICATION GUIDANCE

THIS SECTION IS MANDATORY.

A. TARGET JOB ROLES

Suggest 5 job roles related to the job description.

B. WHERE TO FIND JOBS

Suggest free/commonly used job-search platforms such as:
- LinkedIn
- Naukri
- Indeed
- Wellfound
- Company career pages

C. SEARCH KEYWORDS

Give useful keywords the student can search for on job portals.

D. APPLICATION TIPS

Give 5 practical tips for applying to these jobs.

E. JOB SEARCH STRATEGY

Give a simple step-by-step strategy:

1. Find suitable jobs.
2. Check required skills.
3. Compare them with the resume.
4. Customize the resume.
5. Apply through the appropriate platform.
6. Track applications.
7. Prepare for interviews.

IMPORTANT:
Do NOT provide fake or invented current job openings.
Only provide job-search guidance based on the job description.

==================================================

FINAL CHECK:

Before answering, verify that sections 1, 2, 3, 4, 5, 6, AND 7
are all present in your response.

Never omit Section 7.
""")
```

```python
chain = prompt | llm

print("CareerMate AI chain created successfully!")
```

```python
response = chain.invoke({
    "resume": resume_text,
    "job_description": job_description
})

print("\n========== CAREERMATE AI ==========\n")
print(response.content)
```

```python
from gtts import gTTS
from IPython.display import Audio, display

# Get CareerMate AI response
career_text = response.content

# Convert text to speech
tts = gTTS(
    text=career_text,
    lang="en",
    slow=False
)

# Save audio
audio_file = "careermate_analysis.mp3"
tts.save(audio_file)

print("🔊 CareerMate AI voice generated successfully!")

# Play audio in Colab
display(Audio(audio_file, autoplay=False))
```
