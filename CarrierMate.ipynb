```python
def create_compact_profile(resume):

    sections = {
        "Skills": "",
        "Education": "",
        "Experience": "",
        "Projects": "",
        "Certifications": ""
    }

    lines = [
        line.strip()
        for line in resume.splitlines()
        if line.strip()
    ]

    current_section = None

    section_names = {
        "skills": "Skills",
        "education": "Education",
        "experience": "Experience",
        "projects": "Projects",
        "certifications": "Certifications"
    }

    for line in lines:

        lower_line = line.lower().strip()

        if lower_line in section_names:
            current_section = section_names[lower_line]
            continue

        if current_section:
            sections[current_section] += line + " "

    profile = f"""
Skills:
{sections["Skills"].strip() or "Not mentioned"}

Education:
{sections["Education"].strip() or "Not mentioned"}

Experience:
{sections["Experience"].strip() or "Not mentioned"}

Projects:
{sections["Projects"].strip() or "Not mentioned"}

Certifications:
{sections["Certifications"].strip() or "Not mentioned"}
"""

    return profile.strip()


compact_profile = create_compact_profile(resume_text)

print("========== COMPACT PROFILE ==========")
print(compact_profile)
```

```python
import gradio as gr


def chat_with_careermate(
    user_message,
    history
):

    if not user_message or not user_message.strip():

        return (
            history,
            "",
            None
        )


    # Get CareerMate response
    answer = careermate_response(
        user_message=user_message,
        compact_profile=compact_profile,
        job_description=job_description
    )


    # Add user message
    history.append({
        "role": "user",
        "content": user_message
    })


    # Add CareerMate response
    history.append({
        "role": "assistant",
        "content": answer
    })


    # Generate voice
    audio_path = generate_careermate_voice(answer)


    return (
        history,
        "",
        audio_path
    )
```

```python
initial_message = {
    "role": "assistant",
    "content": "What's your target role?"
}


with gr.Blocks(
    title="CareerMate AI"
) as demo:


    gr.Markdown("""
    # 🤖 CareerMate AI

    ### Your Brutally Honest AI Career Mentor

    **No excuses. No fake motivation. Prove your skills.**

    *Warning: CareerMate has zero patience for weak answers.*
    """)


    chatbot = gr.Chatbot(
        value=[initial_message],
        label="CareerMate",
        height=550
    )


    audio_output = gr.Audio(
        label="🔊 CareerMate Voice",
        autoplay=True
    )


    with gr.Row():

        message = gr.Textbox(
            placeholder="Type your answer...",
            show_label=False,
            scale=8,
            lines=1
        )


        send = gr.Button(
            "➤",
            scale=1
        )


    send.click(
        chat_with_careermate,
        inputs=[
            message,
            chatbot
        ],
        outputs=[
            chatbot,
            message,
            audio_output
        ]
    )


    message.submit(
        chat_with_careermate,
        inputs=[
            message,
            chatbot
        ],
        outputs=[
            chatbot,
            message,
            audio_output
        ]
    )


demo.launch(
    share=True,
    debug=True
)
```
