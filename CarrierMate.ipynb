```python
import os
import getpass
from google.colab import files

from pypdf import PdfReader

from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate

print("Libraries loaded successfully!")
```

```python
os.environ["GROQ_API_KEY"] = getpass.getpass(
    "Enter your Groq API key: "
)

print("Groq API key added successfully!")
```

```python
uploaded = files.upload()
```

```python
resume_filename = list(uploaded.keys())[0]

reader = PdfReader(resume_filename)

resume_text = ""

for page in reader.pages:
    text = page.extract_text()

    if text:
        resume_text += text + "\n"

print("Resume loaded successfully!")
print("Characters extracted:", len(resume_text))
```

```python
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0.2
)

print("Groq model ready!")
```

```python
job_description = """
We are looking for an AI/ML Engineer.

Required skills:
Python
Machine Learning
Deep Learning
SQL
TensorFlow
Data Analysis
Generative AI
"""

print(job_description)
```

```python
prompt = ChatPromptTemplate.from_template("""
You are CareerMate AI, an intelligent career assistant.

Analyze the student's resume against the job description.

RESUME:
{resume}

JOB DESCRIPTION:
{job_description}

Provide the result in the following format:

1. RESUME SUMMARY
Give a short summary of the candidate.

2. SKILLS FOUND
List the technical and soft skills found in the resume.

3. MATCHING SKILLS
List skills present in both the resume and job description.

4. MISSING SKILLS
List important job skills that are missing from the resume.

5. INTERVIEW QUESTIONS
Generate 5 technical interview questions based on the candidate's resume
and the job description.

6. LEARNING ROADMAP
Suggest a simple 30-day learning roadmap for the missing skills.

Keep the explanation simple and student-friendly.
""")
```

```python
chain = prompt | llm

print("CareerMate AI chain created successfully!")
```

```python
response = chain.invoke({
    "resume": resume_text,
    "job_description": job_description
})

print("\n========== CAREERMATE AI ==========\n")
print(response.content)
```
