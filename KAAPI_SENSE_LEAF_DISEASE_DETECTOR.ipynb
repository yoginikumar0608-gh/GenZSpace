```python
%%writefile app.py
import streamlit as st
from transformers import pipeline
from PIL import Image
from gtts import gTTS

classifier = pipeline("image-classification", model="nateraw/vit-base-beans")
advice = {
    "angular_leaf_spot": "Signs of angular leaf spot detected. Avoid overhead watering and apply a fungicide.",
    "bean_rust": "Signs of rust detected. Remove affected leaves and apply a sulfur-based fungicide.",
    "healthy": "This leaf looks healthy. No action needed.",
}

st.title("Kaapi Sense - Leaf Disease Detector")
uploaded = st.file_uploader("Upload a leaf photo")
if uploaded:
    img = Image.open(uploaded).convert("RGB")
    result = classifier(img)
    st.write(result)
    label = result[0]["label"]
    advice_text = advice.get(label, "Consult a local agricultural officer for this condition.")
    st.write(advice_text)
    gTTS(advice_text).save("output.mp3")
    st.audio("output.mp3")
```
